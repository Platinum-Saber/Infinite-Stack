
## 3) Singly Linked List (Python)

**C++ Reference (singly list node):**
```cpp
struct Node {
    int val;
    Node* next;
    Node(int v): val(v), next(nullptr) {}
};
```


````python

class Node:
    __slots__ = ("val", "next")
    def __init__(self, val, nxt=None):
        self.val = val
        self.next = nxt

class SinglyList:
    def __init__(self):
        self.head = None

    def push_front(self, v):
        self.head = Node(v, self.head)

    def find(self, target):
        cur = self.head
        while cur:
            if cur.val == target:
                return cur
            cur = cur.next
        return None

    def delete_value(self, target):
        dummy = Node(0, self.head)
        prev, cur = dummy, self.head
        while cur:
            if cur.val == target:
                prev.next = cur.next
                break
            prev, cur = cur, cur.next
        self.head = dummy.next

L = SinglyList()
for x in [3,1,4,1,5]:
    L.push_front(x)
assert L.find(4) is not None
L.delete_value(1)
"deleted one 1; head now -> {}".format(L.head.val)
````

## 5) Sorting


**C++ Reference (Merge Sort function):**
```cpp
void merge(vector<int>& a, int l, int m, int r){
    vector<int> L(a.begin()+l, a.begin()+m+1);
    vector<int> R(a.begin()+m+1, a.begin()+r+1);
    int i=0,j=0,k=l;
    while(i<L.size() && j<R.size())
        a[k++] = (L[i]<=R[j]) ? L[i++] : R[j++];
    while(i<L.size()) a[k++] = L[i++];
    while(j<R.size()) a[k++] = R[j++];
}
void mergesort(vector<int>& a, int l, int r){
    if(l>=r) return;
    int m=(l+r)/2;
    mergesort(a,l,m);
    mergesort(a,m+1,r);
    merge(a,l,m,r);
}
```


````python

def merge_sort(a):
    if len(a) <= 1:
        return a
    mid = len(a)//2
    left = merge_sort(a[:mid])
    right = merge_sort(a[mid:])
    i=j=0
    out=[]
    while i<len(left) and j<len(right):
        if left[i] <= right[j]:
            out.append(left[i]); i+=1
        else:
            out.append(right[j]); j+=1
    out.extend(left[i:]); out.extend(right[j:])
    return out

merge_sort([5,2,9,1,5,6])
````

````python

def quick_sort(a):
    if len(a) <= 1:
        return a
    pivot = a[len(a)//2]
    lt = [x for x in a if x < pivot]
    eq = [x for x in a if x == pivot]
    gt = [x for x in a if x > pivot]
    return quick_sort(lt) + eq + quick_sort(gt)

quick_sort([3,6,8,10,1,2,1])
````

## 6) Searching

````python

def linear_search(a, target):
    for i, v in enumerate(a):
        if v == target:
            return i
    return -1

def binary_search(a, target):
    lo, hi = 0, len(a)-1
    while lo <= hi:
        mid = (lo+hi)//2
        if a[mid] == target:
            return mid
        if a[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

arr = sorted([5,1,8,3,2,9])
linear_search(arr, 8), binary_search(arr, 8)
````

## 7) Binary Search Tree (BST)


**C++ Reference (BST insert & inorder):**
```cpp
struct Node{int key; Node* l; Node* r; Node(int k):key(k),l(nullptr),r(nullptr){} };
Node* insert(Node* root, int k){
    if(!root) return new Node(k);
    if(k < root->key) root->l = insert(root->l, k);
    else if(k > root->key) root->r = insert(root->r, k);
    return root;
}
void inorder(Node* root){
    if(!root) return;
    inorder(root->l);
    cout << root->key << " ";
    inorder(root->r);
}
```


````python

class BST:
    class N:
        __slots__=("k","l","r")
        def __init__(self,k): self.k=k; self.l=None; self.r=None
    def __init__(self): self.root=None
    def insert(self,k):
        def rec(n,k):
            if not n: return BST.N(k)
            if k<n.k: n.l=rec(n.l,k)
            elif k>n.k: n.r=rec(n.r,k)
            return n
        self.root=rec(self.root,k)
    def inorder(self):
        out=[]
        def rec(n):
            if not n: return
            rec(n.l); out.append(n.k); rec(n.r)
        rec(self.root); return out

t=BST()
for x in [5,3,7,2,4,6,8]:
    t.insert(x)
t.inorder()
````

## 9) Graph Traversal (BFS/DFS)

````python

from collections import deque, defaultdict

def bfs(adj, start):
    vis = {start}
    q = deque([start])
    order = []
    while q:
        u = q.popleft()
        order.append(u)
        for v in adj[u]:
            if v not in vis:
                vis.add(v); q.append(v)
    return order

def dfs(adj, start):
    vis=set(); order=[]
    def rec(u):
        vis.add(u); order.append(u)
        for v in adj[u]:
            if v not in vis: rec(v)
    rec(start); return order

adj = defaultdict(list, {
    0:[1,2], 1:[2], 2:[0,3], 3:[3]
})
bfs(adj, 2), dfs(adj, 2)
````

## 10) Dijkstra Shortest Path

````python

import heapq
from collections import defaultdict

def dijkstra(n, edges, src):
    g = defaultdict(list)
    for u,v,w in edges:
        g[u].append((v,w))
        g[v].append((u,w))  # undirected; remove if directed
    dist = [float('inf')]*n
    dist[src]=0
    pq=[(0,src)]
    while pq:
        d,u = heapq.heappop(pq)
        if d!=dist[u]: continue
        for v,w in g[u]:
            nd=d+w
            if nd<dist[v]:
                dist[v]=nd
                heapq.heappush(pq,(nd,v))
    return dist

dijkstra(5, [(0,1,2),(1,2,3),(0,3,1),(3,4,4),(2,4,1)], 0)
````

## 11) Disjoint Set Union (Union‑Find)

````python

class DSU:
    def __init__(self, n):
        self.p=list(range(n))
        self.r=[0]*n
    def find(self,x):
        if self.p[x]!=x:
            self.p[x]=self.find(self.p[x])
        return self.p[x]
    def union(self,a,b):
        ra, rb = self.find(a), self.find(b)
        if ra==rb: return False
        if self.r[ra]<self.r[rb]: ra,rb = rb,ra
        self.p[rb]=ra
        if self.r[ra]==self.r[rb]: self.r[ra]+=1
        return True

dsu = DSU(5)
dsu.union(0,1); dsu.union(1,2); (dsu.find(0)==dsu.find(2))
````

## 12) Dynamic Programming (0/1 Knapsack)

````python

def knapsack_01(weights, values, W):
    n=len(weights)
    dp=[[0]*(W+1) for _ in range(n+1)]
    for i in range(1,n+1):
        w=weights[i-1]; val=values[i-1]
        for cap in range(W+1):
            dp[i][cap]=dp[i-1][cap]
            if w<=cap:
                dp[i][cap]=max(dp[i][cap], dp[i-1][cap-w]+val)
    return dp[n][W]

knapsack_01([2,3,4,5],[3,4,5,8], 5)
````
