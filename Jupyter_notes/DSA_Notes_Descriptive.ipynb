## Part I — Data Structures


### 1) Arrays / Dynamic Arrays

**Definition:** Contiguous memory storing elements of same type; dynamic arrays can resize.

**Operations & Complexity**

| Operation | Time |
|---|---:|
| Random access `a[i]` | `O(1)` |
| Append (amortized) | `O(1)` |
| Insert/Delete middle | `O(n)` |
| Search (unsorted) | `O(n)` |
| Search (sorted, binary) | `O(log n)` |



**Advantages**
- `O(1)` random access.
- Great cache locality.
- Simple & widely available.

**Disadvantages**
- Middle insert/delete `O(n)`.
- Resizing may copy elements.
- Requires contiguous memory.


````python

# Python
a=[3,1,4]
a.append(1)      # amortized O(1)
x=a[2]           # O(1)
print(x)
a.insert(1,42)   # O(n)
a
````


**C++**
```cpp

v.insert(v.begin()+1, 42);      // O(n)
```



### 2) Linked Lists (Singly)

**Definition:** Nodes store a value and a pointer to the next node; not contiguous.



**Operations & Complexity**

| Operation | Time |
|---|---:|
| Insert/Delete at head | `O(1)` |
| Search by value | `O(n)` |
| Access by index | `O(n)` |



**Advantages**
- `O(1)` head insert/delete.
- Flexible growth, no contiguous memory.
- Useful for adjacency lists / queues.

**Disadvantages**
- No random access; `O(n)` indexing.
- Pointer overhead; poor cache locality.
- Reverse traversal requires extra pointers.


````python

# Python
class Node:
    def __init__(self, val, nxt=None):
        self.val, self.next = val, nxt

class SinglyList:
    def __init__(self): self.head=None
    def push_front(self, v): self.head = Node(v, self.head)
    def find(self, t):
        cur=self.head
        while cur and cur.val!=t: cur=cur.next
        return cur
    def delete_value(self, t):
        dummy=Node(0, self.head); prev,cur=dummy,self.head
        while cur:
            if cur.val==t: prev.next=cur.next; break
            prev,cur=cur,cur.next
        self.head=dummy.next

L = SinglyList()
for x in [3,1,4,1,5]: L.push_front(x)
assert L.find(4) is not None
L.delete_value(1)
"head:", L.head.val
````


### 6) Trees — Binary Search Tree (BST)

**Definition:** Binary tree with order Left < Root < Right.



**Complexity**

| Operation | Avg | Worst |
|---|---:|---:|
| Search/Insert/Delete | `O(log n)` | `O(n)` |



**Advantages**
- Ordered keys; in-order traversal sorted.
- Range queries are natural.

**Disadvantages**
- Can degenerate without balancing.
- Use AVL/RB trees for guaranteed `O(log n)`.


````python

# Python
class BST:
    class N:
        def __init__(self,k): self.k, self.l, self.r = k, None, None
    def __init__(self): self.root=None
    def insert(self,k):
        def rec(n,k):
            if not n: return BST.N(k)
            if k<n.k: n.l=rec(n.l,k)
            elif k>n.k: n.r=rec(n.r,k)
            return n
        self.root=rec(self.root,k)
    def inorder(self):
        out=[]
        def rec(n):
            if not n: return
            rec(n.l); out.append(n.k); rec(n.r)
        rec(self.root); return out

t=BST()
for x in [5,3,7,2,4,6,8]: t.insert(x)
t.inorder()
````


**C++**
```cpp

Node* insert(Node* root, int k){
    if(!root) return new Node(k);
    if(k<root->key) root->l=insert(root->l,k);
    else if(k>root->key) root->r=insert(root->r,k);
    return root;
}
```



### 7) Tries (Prefix Trees)

**Definition:** Keys are strings; edges labeled by characters; path spells a key.



**Complexity**
- Insert/Find: `O(L)` with `L` = key length.



**Advantages**
- Fast prefix queries (autocomplete).
- Can annotate counts, prefixes.

**Disadvantages**
- Higher memory usage.
- Larger constants than hashing for single lookups.


````python

# Python
class Trie:
    def __init__(self): self.t = {}
    def insert(self, s):
        node=self.t
        for ch in s: node=node.setdefault(ch,{})
        node['#']=True
    def startswith(self,p):
        node=self.t
        for ch in p:
            if ch not in node: return False
            node=node[ch]
        return True

tr=Trie(); tr.insert("algo"); tr.insert("algae"); tr.startswith("alg"), tr.startswith("alt")
````


### 9) Disjoint Set Union (Union‑Find)

Tracks connectivity with path compression + union by rank/size.



**Complexity:** Amortized almost `O(1)` per op (inverse Ackermann).

**Advantages**
- Excellent for dynamic connectivity, cycle detection, Kruskal MST.

**Disadvantages**
- Limited operations (no order or detailed queries).


````python

# Python
class DSU:
    def __init__(self,n):
        self.p=list(range(n)); self.r=[0]*n
    def find(self,x):
        if self.p[x]!=x: self.p[x]=self.find(self.p[x])
        return self.p[x]
    def union(self,a,b):
        ra,rb=self.find(a),self.find(b)
        if ra==rb: return False
        if self.r[ra]<self.r[rb]: ra,rb=rb,ra
        self.p[rb]=ra
        if self.r[ra]==self.r[rb]: self.r[ra]+=1
        return True

dsu=DSU(5); dsu.union(0,1); dsu.union(1,2); dsu.find(0)==dsu.find(2)
````

## Part II — Algorithms


### 1) Searching

**Linear Search**
- **Idea:** scan sequentially until found.
- **Time:** `O(n)`.
- **Pros:** Works on any container; no preconditions.
- **Cons:** Slow for large `n`.



**Binary Search (sorted array required)**
- **Idea:** halve the search space each step.
- **Time:** `O(log n)`.
- **Pros:** Very fast; minimal memory.
- **Cons:** Requires sorted data + random access; off‑by‑one pitfalls.


````python

# Python
def linear_search(a,t):
    for i,v in enumerate(a):
        if v==t: return i
    return -1

def binary_search(a,t):
    lo,hi=0,len(a)-1
    while lo<=hi:
        mid=(lo+hi)//2
        if a[mid]==t: return mid
        if a[mid]<t: lo=mid+1
        else: hi=mid-1
    return -1

arr=sorted([5,1,8,3,2,9])
linear_search(arr,8), binary_search(arr,8)
````


### 2) Sorting — Trade‑offs

| Algorithm | Best | Avg | Worst | Stable | In‑place | Pros | Cons |
|---|---:|---:|---:|:---:|:---:|---|---|
| Insertion | `n` | `n²` | `n²` | ✅ | ✅ | Great for small/near‑sorted | Poor on large random data |
| Merge | `n log n` | `n log n` | `n log n` | ✅ | ❌ | Predictable, stable | Extra memory |
| Heap | `n log n` | `n log n` | `n log n` | ❌ | ✅ | In‑place, good worst‑case | Larger constants |
| Quick | `n log n` | `n log n` | `n²` | ❌ | ✅ | Fast in practice | Worst‑case `n²` |
| Counting/Radix | `n+k` | `n+k` | `n+k` | ✅ | ❌ | Linear for small integer ranges | Needs constraints |


````python

# Python — Merge Sort
def merge_sort(a):
    if len(a)<=1: return a
    mid=len(a)//2
    L=merge_sort(a[:mid]); R=merge_sort(a[mid:])
    i=j=0; out=[]
    while i<len(L) and j<len(R):
        if L[i]<=R[j]: out.append(L[i]); i+=1
        else: out.append(R[j]); j+=1
    out.extend(L[i:]); out.extend(R[j:])
    return out

merge_sort([5,2,9,1,5,6])
````

````python

# Python — Quick Sort (simple)
def quick_sort(a):
    if len(a)<=1: return a
    p=a[len(a)//2]
    return quick_sort([x for x in a if x<p]) + [x for x in a if x==p] + quick_sort([x for x in a if x>p])

quick_sort([3,6,8,10,1,2,1])
````


### 3) Graph Traversals (BFS/DFS)

**BFS**
- Use: unweighted shortest paths, levels.
- Time: `O(V+E)`.
- Pros: finds fewest-edge paths.
- Cons: can use `O(V)` memory.

**DFS**
- Use: components, topological sort, cycle detection.
- Time: `O(V+E)`.
- Pros: simple recursion/stack.
- Cons: deep recursion risk.


````python

from collections import deque, defaultdict

def bfs(adj, start):
    vis={start}; q=deque([start]); order=[]
    while q:
        u=q.popleft(); order.append(u)
        for v in adj[u]:
            if v not in vis:
                vis.add(v); q.append(v)
    return order

def dfs(adj, start):
    vis=set(); order=[]
    def rec(u):
        vis.add(u); order.append(u)
        for v in adj[u]:
            if v not in vis: rec(v)
    rec(start); return order

adj=defaultdict(list, {0:[1,2], 1:[2], 2:[0,3], 3:[3]})
bfs(adj,2), dfs(adj,2)
````


### 4) Shortest Path — Dijkstra (non‑negative weights)

- Idea: greedily expand nearest node using a min‑heap.
- Time: `O((V+E) log V)` with heap.
- Pros: fast on sparse graphs.
- Cons: not valid with negative edges.


````python

import heapq, collections
def dijkstra(n, edges, src):
    g=collections.defaultdict(list)
    for u,v,w in edges:
        g[u].append((v,w))
        g[v].append((u,w))  # undirected; remove if directed
    INF=float('inf'); dist=[INF]*n; dist[src]=0
    pq=[(0,src)]
    while pq:
        d,u=heapq.heappop(pq)
        if d!=dist[u]: continue
        for v,w in g[u]:
            nd=d+w
            if nd<dist[v]: dist[v]=nd; heapq.heappush(pq,(nd,v))
    return dist

dijkstra(5, [(0,1,2),(1,2,3),(0,3,1),(3,4,4),(2,4,1)], 0)
````


### 5) Minimum Spanning Tree — Kruskal

- Idea: sort edges by weight; add if endpoints are in different components (DSU).
- Time: `O(E log E)`.
- Pros: simple with edge lists; excellent with DSU.
- Cons: requires sorting; Prim may be preferable on dense graphs.


````python

class DSU2:
    def __init__(self,n):
        self.p=list(range(n)); self.r=[0]*n
    def find(self,x):
        if self.p[x]!=x: self.p[x]=self.find(self.p[x])
        return self.p[x]
    def union(self,a,b):
        ra,rb=self.find(a),self.find(b)
        if ra==rb: return False
        if self.r[ra]<self.r[rb]: ra,rb=rb,ra
        self.p[rb]=ra
        if self.r[ra]==self.r[rb]: self.r[ra]+=1
        return True

def mst_kruskal(n, edges):
    # edges: (w,u,v)
    edges=sorted(edges)
    dsu=DSU2(n); total=0; used=[]
    for w,u,v in edges:
        if dsu.union(u,v):
            total+=w; used.append((u,v,w))
    return total, used

mst_kruskal(4, [(1,0,1),(4,0,2),(3,1,2),(2,1,3),(5,2,3)])
````


### 6) Dynamic Programming — 0/1 Knapsack

- Idea: DP over (i, capacity): choose take/skip for each item.
- Time: `O(nW)`. Space: `O(nW)` or `O(W)` optimized.
- Pros: exact solution.
- Cons: pseudo‑polynomial in capacity `W`.


````python

def knapsack_01(W, wt, val):
    n=len(wt)
    dp=[[0]*(W+1) for _ in range(n+1)]
    for i in range(1,n+1):
        for cap in range(W+1):
            dp[i][cap]=dp[i-1][cap]
            if wt[i-1]<=cap:
                dp[i][cap]=max(dp[i][cap], dp[i-1][cap-wt[i-1]]+val[i-1])
    return dp[n][W]

knapsack_01(5, [2,3,4,5], [3,4,5,8])
````


### 7) Patterns: Greedy, Sliding Window, Two Pointers

**Greedy**
- Pros: very fast; often near‑linear.
- Cons: correctness requires proof (greedy‑choice + optimal substructure).

**Sliding Window**
- Pros: converts nested loops to linear scans when answer is a contiguous subarray/window.
- Cons: not applicable if solution isn't contiguous.

**Two Pointers**
- Pros: linear solutions on sorted/monotone data.
- Cons: requires structure; not universal.


````python

def longest_unique_substring(s):
    seen={}; start=0; best=0
    for i,ch in enumerate(s):
        if ch in seen and seen[ch]>=start:
            start=seen[ch]+1
        seen[ch]=i
        best=max(best, i-start+1)
    return best

longest_unique_substring("abcbcbb")
````
